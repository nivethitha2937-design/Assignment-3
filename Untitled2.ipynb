```python
from google import genai
from google.genai import types

# Initialize your Gemini client
client = genai.Client()

knowledge_base = """
COLLEGE CANTEEN / CAFE: Campus Cafe

Q: Where is the canteen located?
A: The canteen is located inside the college campus.

Q: How can I order food?
A: You can order food directly at the canteen counter.

Q: What food items are available?
A: We offer sandwiches, burgers, pizza, dosa, idli, fried rice,
noodles, snacks, and other food items.

Q: What beverages are available?
A: We offer tea, coffee, fresh juices, milkshakes, and soft drinks.

Q: Do you have vegetarian food?
A: Yes, vegetarian food options are available.

Q: Do you have non-vegetarian food?
A: Yes, non-vegetarian food options are available.

Q: What is the price of food?
A: Prices vary depending on the food item. Please check the menu for details.

Q: Is the food affordable for students?
A: Yes, the cafe offers affordable food options for college students.

Q: Can I get takeaway?
A: Yes, takeaway is available.

Q: Is seating available?
A: Yes, seating is available for students and staff.

Q: What are the opening hours?
A: The canteen is open during college working hours.

Q: Is the canteen open during lunch break?
A: Yes, the canteen serves food during student break times.

Q: What payment methods are accepted?
A: Cash and UPI payments are accepted.

Q: Do you offer combo meals?
A: Combo meals may be available. Please check with the canteen staff.

Q: Do you have special offers?
A: Special offers may be available from time to time.

Q: Can I order food for a group?
A: You can enquire at the counter about group orders.

Q: Can I arrange a small celebration?
A: Please contact the canteen staff for celebration arrangements.

Q: Who can use the canteen?
A: Students, staff, and authorized college members can use the canteen.

Q: How can I get more information?
A: You can visit the canteen counter for further assistance.
"""

system_instruction = f"""
You are a friendly, professional customer support assistant for a College Canteen/Cafe.

KNOWLEDGE BASE:
{knowledge_base}

RULES:
1. Always use the knowledge base to answer questions.
2. Match questions by meaning, even when the wording is different
   or the user makes spelling or grammar mistakes.
3. Answer directly when the information is available.
4. Use only information supported by the knowledge base.
5. Do not invent prices, services, policies, or other details.
6. If the answer is not available, politely explain that you don't
   have that information in the knowledge base.
7. Keep answers concise, clear, and natural.
8. Do not refuse a question when its answer is available above.
"""

# Create the chat
chat = client.chats.create(
    model="gemini-3.1-flash-lite",
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.2,
        max_output_tokens=512
    ),
    history=[]
)

print("Hi! I'm College Canteen Support. How can I help you?")
print("Type 'quit' to exit.\n")

while True:
    user_input = input("You: ").strip()

    if user_input.lower() in ["quit", "exit", "bye"]:
        print("College Canteen Support: Thanks for reaching out. Have a great day!")
        break

    if not user_input:
        continue

    try:
        response = chat.send_message(user_input)

        if response.text:
            print(f"AI: {response.text}\n")
        else:
            print("AI: Sorry, I couldn't generate a response. Please try again.\n")

    except Exception as e:
        print(f"Error generating response: {e}\n")
```
